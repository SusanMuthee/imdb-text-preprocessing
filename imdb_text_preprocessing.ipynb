```python
# import necessary libraries
import numpy as np
import pandas as pd 
import re
from bs4 import BeautifulSoup
import nltk 
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
# load data and inspect 
df = pd.read_csv("C:/Users/u1172021/Downloads/IMDB Dataset.csv/IMDB Dataset.csv")
df.head()                         # display first 5 rows
df.info()                         # col names, data types, non-null counts
df.shape                          # total rows and columns
df['sentiment'].value_counts()    # checks class balance
df.isnull().sum()                 # check for missing values
df.duplicated().sum()             # check for duplicated rows
```

```python
text = df['review']    # define text
print(text[:200])      # view last 200 raw text before cleaning
```

```python
# remove HTML tags
df['review_wo_html'] = df['review'].apply(lambda x: BeautifulSoup(x, "html.parser").get_text())
df['review_wo_html'].head()
```

```python
# lowercase
df['review_lower'] = df['review_wo_html'].str.lower()
df['review_lower'].head()
```

```python
# remove punctuation and special characters
df['review_wo_punct'] = df['review_lower'].apply(lambda x: re.sub(r"[^a-z\s]", " ", x))
df['review_wo_punct'].head()
```

```python
# tokenization
df['review_tokens'] = df['review_wo_punct'].apply(lambda x: word_tokenize(x))
df['review_tokens'].head()
```

```python
# remove stopwords
# define stopwords
stop_words = set(stopwords.words('english'))

# remove stopwords
df['review_wo_stopwords'] = df['review_tokens'].apply(lambda x: [t for t in x if t not in stop_words])
df['review_wo_stopwords'].head()
```

```python
# lemmatization 
# define lemmatizer
lemmatizer = WordNetLemmatizer()

# lemmatize
df['review_lemmatized'] = df['review_wo_stopwords'].apply(lambda x: [lemmatizer.lemmatize(t) for t in x])
df['review_lemmatized'].head()
```

```python
# join tokens back into a string 
df['review_clean'] = df['review_lemmatized'].apply(lambda x: " ".join(x))
df['review_clean'].head()
```

```python
# text preprocessing function 
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess(text):

    # remove HTML tags
    text = BeautifulSoup(text, "html.parser").get_text()

    # lowercase 
    text = text.lower()

    # remove punctuation and special characters
    text = re.sub(r"[^a-z\s]", " ", text)

    # tokenize 
    tokens = word_tokenize(text)

    # remove stopwords
    tokens = [t for t in tokens if t not in stop_words]

    # lemmatize
    text = [lemmatizer.lemmatize(t) for t in tokens]

    
    return tokens

# apply function to reviews
df['review_processed'] = df['review'].apply(preprocess)
df['review_processed'].head()
```
